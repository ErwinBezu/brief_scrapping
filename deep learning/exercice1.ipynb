## Porte AND

```python
import numpy as np

X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([0, 0, 0, 1])                  # AND ; pour NOR : np.array([1, 0, 0, 0])

np.random.seed(42)
w = np.random.randn(2)
b = np.random.randn()

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def predict(X, w, b):
    return sigmoid(X @ w + b)

def bce(y, p, eps=1e-12):                   # entropie croisée binaire
    return -np.mean(y * np.log(p + eps) + (1 - y) * np.log(1 - p + eps))

lr, epochs = 0.5, 2000
for i in range(epochs):
    p = predict(X, w, b)
    dz = p - y                              # gradient de l'entropie croisée
    w -= lr * (X.T @ dz) / len(X)
    b -= lr * dz.mean()
    if i % 200 == 0:
        print(f"Epoch {i}, Loss: {bce(y, p):.4f}")

print("Prédictions :", predict(X, w, b).round(3))
```
