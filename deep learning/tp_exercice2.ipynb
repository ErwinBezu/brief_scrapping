```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import networkx as nx
import tensorflow as tf
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input

tf.keras.utils.set_random_seed(42)
```

```python
iris = load_iris()
X = iris.data
y = iris.target

print("Dimensions X :", X.shape)
print("Dimensions y :", y.shape)
```

```python
print("Exemples de X :\n", X[:5])
print("Exemples de y :\n", y[:5])
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size= 0.2, stratify=y, random_state=42
)

print("Taille train :", X_train.shape)
print("Taille test :", X_test.shape)
```

```python
encoder =  OneHotEncoder(sparse_output=False)
y_train = encoder.fit_transform(y_train.reshape(-1, 1))
y_test = encoder.transform(y_test.reshape(-1, 1))
print("Sorties encodées :\n", y_train[:5])
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
print("Features normalisées :\n", X_train[:5])
```

```python
model = Sequential([
    Input(shape=(4,)),
    Dense(8, activation='relu'),
    Dense(3, activation='softmax')
])
```

```python
model.summary()
```

```python
layers = [4, 8, 3]

G = nx.DiGraph()
pos = {}
layer_positions = []

for l, n_neurons in enumerate(layers):
    layer_pos = []
    for i in range(n_neurons):
        node_id = f"L{l}N{i}"
        G.add_node(node_id, layer=l)
        pos[node_id] = (l, -i)
        layer_pos.append(node_id)
    layer_positions.append(layer_pos)

for l in range(len(layers)-1):
    for src in layer_positions[l]:
        for dst in layer_positions[l+1]:
            G.add_edge(src, dst)

plt.figure(figsize=(8, 6))
nx.draw(G, pos, with_labels=False, 
        node_size=500, node_color="lightgreen", edge_color="gray")
plt.title("Réseau de neurones pour Iris (4-8-3)", fontsize=14)
plt.show()        
```

```python
model.compile(
    optimizer='adam', 
    loss='categorical_crossentropy', 
    metrics=['accuracy']
)
```

```python
history = model.fit(X_train, y_train, epochs=100, batch_size=5, validation_split=0.1)
```

model.fit() : Lance l'entraînement du réseau de neurones.   
X_train : Les données utilisées pour apprendre (les caractéristiques)  
y_train : Les réponses attendues correspondant aux données d'entraînement.  
epochs=100 : Le modèles effectue 100 passages complets sur les données utilisées pour l'apprentissage.    
batch_size=5 : Le modèle traite les données par groupes de 5 exemples avant de mettre à jour ses poids.   
validation_split=0.1 : Réserve 10% des données d'entraînement pour vérifier les performances du modèles pendant l'apprentissage.   
history : Conserve l'historique de l'entraînement, notammennt la loss et l'accuracy

```python
loss, accuracy = model.evaluate(X_test, y_test)
print(f"Test Accuracy: {accuracy*100:.2f}%")
```

```python
predictions = model.predict(X_test)
print("Prédictions :\n", predictions[:5].round(3))
print("Classes prédites :", np.argmax(predictions, axis=1)[:20])
print("Vraies classes : ", np.argmax(y_test, axis=1)[:20])
```

```python
y_true = np.argmax(y_test, axis=1)
y_pred = np.argmax(predictions, axis=1)

df = pd.DataFrame(scaler.inverse_transform(X_test), columns=iris.feature_names).round(1)

df["True_Class"] = y_true
df["Predicted_Class"] = y_pred

df.head(10)
```
