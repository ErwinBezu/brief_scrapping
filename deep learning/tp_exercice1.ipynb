```python
import numpy as np
import pandas as pd
```

```python
X = np.array([[0,0],
             [0,1],
             [1,0],
             [1,1]])

# y = np.array([0, 0, 0, 1])

y = np.array([1, 0, 0, 0])
```

```python
df_X = pd.DataFrame(X, columns = ['Feature_1', 'Feature_2'])
print("DataFrame X :")
df_X
```

```python
df_y = pd.DataFrame(y, columns=['Target'])
print("\nDataFrame y :")
df_y
```

```python
np.random.seed(42)
w = np.random.randn(2)
b = np.random.randn()
print("Poids initiaux :", w)
print("Bias initial :", b)
```

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

```python
def predict(X, w, b):
    z = np.dot(X, w) + b
    return sigmoid(z)
```

```python
def loss(y_true, y_pred):
    return np.mean((y_true - y_pred) ** 2)
```

```python
learning_rate = 0.1
epochs = 2000

for i in range (epochs):
    y_pred = predict(X, w, b)
    dw = -2 * np.dot((y - y_pred) * y_pred * ( 1 - y_pred), X)  / len(X)
    db = -2 * np.mean((y - y_pred) * y_pred * ( 1 - y_pred))

    w -= learning_rate * dw
    b -= learning_rate * db

    if i % 100 == 0:
        print(f"Epoch {i}, Loss: {loss(y, y_pred)}")
```

```python
y_pred =  predict(X, w, b)
print("Prédictions finales :", y_pred)
print("Classes prédites (seuil 0.5) :", (y_pred > 0.5).astype(int))
```

```python
np.random.seed(42)
w = np.random.randn(2)
b = np.random.randn()

def bce(y_true, y_pred, eps=1e-12):
    return -np.mean(y_true * np.log(y_pred + eps) + (1 - y_true) * np.log(1 - y_pred + eps))

learning_rate = 0.5
for i in range(2000) : 
    y_pred = predict(X, w, b)
    dz =  y_pred - y
    w -= learning_rate * np.dot(dz, X) / len(X)
    b -= learning_rate * np.mean(dz)
    if i % 400 == 0:
        print(f"Epoch {i}, Loss: {bce(y, y_pred): .4f}")

print("Prédictions finales :", predict(X, w, b).round(3))
```
